## Pandas Series
A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.

## Creating Series
There are multiple ways to create a pandas Series:

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a','b','c']
list = [10,20,30]
arr = np.array([2,4,6])
d = {1:10,2:20,3:30}
```

```python
pd.Series(list) #labels and data points are result
```

```python
pd.Series(list, index = labels) #for setting custom labels/indexes
```

```python
pd.Series(arr, arr) #can also send 1-dim arr
```

```python
pd.Series(d)
```
