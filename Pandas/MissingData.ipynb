```python
import numpy as np
import pandas as pd
```

**Finding Missing Data**

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
#nan is Not A Number
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum() # sum of number of NA values
```

```python
df.isna().any() # checks if any column has atleast 1 na value
```

**Removing Missing Data**

```python
df
```

```python
df.dropna() # drop na works on rows. drops all the rows having atleat 1 na..if no threshold is provided
```

```python
df
```

```python
df.dropna(thresh = 3) #keeps rows having atleast 3 NON-NA value
```

**Filling the missing Data**

```python
df
```

```python
df.fillna(0) 
# just an example.. to fix the na value use inplace
```

```python
values = {
    'A': 100,
    'B': 200,
    'C': 300,
    'D': 400
}
df.fillna(value = values)
# fill hoga COLUMN WISE
# drop and select hoga ROW WISE
```

```python
df.fillna(df.mean())
```
