```python
import numpy as np 
import pandas as pd
```

**Creating a DataFrame**

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York','Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
column = ['Name', 'Age', 'City', 'Salary']
df2 = pd.DataFrame(data_list, columns = column)

#in lists we dont provide a name that's why in the title we see 0,1,...(default)
```

```python
df2 #hence we can create series with both lists and dic.. but in list we have to name columns manually.
```

**Selecting and Indexing of Columns**

```python
df2
```

```python
df2['Name'] #Selecting a column
```

```python
df2[['Name','City']] #selecting multiple column
```

**Creating a new column**

```python
df2['Designation'] = ['Doctor', 'Eng.', 'Artist', 'Chef']
df2
```

**Removing Columns**

```python
df2.drop('Age', axis = 1) 

# on default its set to 0 so if u dont provide it will give error
# BUT the removed item is not erased from the memory..
```

```python
df2.drop('Designation', axis = 1, inplace = True)
```

```python
df2 #using inplace permanently removes item
```
