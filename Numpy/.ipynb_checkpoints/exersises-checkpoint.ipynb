**Valid Sudoku**

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
r = np.sum(s, axis = 1) #every row ka sun
r
```

```python
c = np.sum(s, axis = 0)
c
```

```python
slice = s[:3, :3]
slice
```

```python
b = slice
b = b.sum()
if b != 45:
  print("Incorrect")
else:
  print("Correct")
```

**General Questions**

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 69],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
data.shape
```

```python
#Find the average age of students.
age = data[:, 0]
age
```

```python
age.sum() / 5
```

```python
#Extract Math marks of all students.
m = data[:, 1]
m
```

```python
#Find the highest Science mark.
s = data[:, 2]
s
```

```python
s.max()
```

```python
#Get details of the student who scored more than 90 in Math.
data[data[:,1]>90]
```

```python
#Increase Math marks of all students by 5.
m
```

```python
m + 5
```

```python
#Find how many students are younger than 19.
age
```

```python
len(age[age<19])
```

```python
#Calculate the average marks in each subject (column-wise mean).
m.mean()
```

```python
s.mean()
```

```python
#Get data of students who scored at least 80 in both subjects.
data[(data[:,1] >= 80) & (data[:,2] >= 80)]
```

```python
#Replace all Science marks < 75 with 0.
s
```

```python
for i in s:
    if i < 75:
        print("0")
    else:
        print(i)
```

```python
data[:,2][data[:,2] < 75] = 0
data[:,2]
```
