**NUMPY INDEXING AND SLICING OF VECTORS**

```python
arr = np.arange(11,21)
arr
```

```python
arr[0]
```

**NUMPY INDEXING AND SLICING OF MATRIX**

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0] #this element (0) represents the row index.. startiing from zero
```

```python
arr[0,0] #this gives us a single element..[row,column]
```

```python
arr[0:2,1:3] #for slicing in matrix.
```

```python
arr[3:,3:]
```

```python
arr[:,2] #for sliciing a column
```

**BOOLEAN INDEXING**

```python
arr = np.arange(11,21)
arr
```

**IF WE WANT ONLY EVEN NUMBERS IN OUR ARRAY THEN WE CAN DO THIS:**

```python
index_bool = arr % 2 == 0 #GIVES TRUE FOR EVEN ELEMENTS
index_bool
```

```python
arr = arr[index_bool] #pass only the true values in the array
arr
```
