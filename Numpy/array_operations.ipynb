_Deep and Shallow copy_

SHALLOW COPY means you slice a part of og array and the sliced part gets different memory location so it forms a **shallow copy** performing operations of sliced array doesnt change the og array

DEEP COPY means you make a copy of the og array so the locations of both array are same and performing operations on the copied array results in changing the og array this is **deep copy**

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice = slice * 10
slice
```

```python
a #shallow copy - no change in og

#for shallow copy we can also use .copy() like: b = a.copy()
```

```python
b = a
b
```

```python
b[0] = 99
b
```

```python
a #now this is deep copy - changed og
```

**ADVANCED ARRAY MANIPULATION**

_Stacking Array_: can only be applied on array/vectors.. cause matrix is already stacked.

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```python
np.vstack((a,b)) #stacks vertically, passing as a tuple
```

```python
np.hstack((a,b)) #stacks horizontally
```

```python
np.column_stack((a,b))
```
